```python
#начало 6 задачи игра жизнь с numpy
def life(life_matrix):
      p = np.pad(life_matrix, 1)
      #print(p)
      n = (p[:-2, :-2] + p[:-2, 1:-1] + p[:-2, 2:] +
           p[1:-1, :-2]               + p[1:-1, 2:] +
           p[2:, :-2]  + p[2:, 1:-1]  + p[2:, 2:])    
      p = ((n == 3) | ((life_matrix == 1) & (n == 2)))
      return(p.astype(int))
```
