```python
import pandas as pd 
import numpy as np
anime = pd.read_csv('anime-dataset-2023.csv')
anime_copy = anime.copy(deep=True)

anime.tail()
genres = (anime['Genres'].str.split(', ').explode())
set(genres)
#anime['Score'].astype(float)
anime['Score'] = pd.to_numeric(anime['Score'], errors='coerce')
```

```python

a = anime.assign(Genres = anime['Genres'].str.split(', ')).explode('Genres').groupby('Genres', as_index = False)['Score'].mean()
print(a)
```
