```python
def live_cells(matrix, i, j):
      lengh, high = matrix.shape
      count = 0
      for x in range(-1, 2):
          for y in range(-1, 2):
              if x == 0 and y == 0:
                  continue
              xi = x + i
              yj = y + j
              if 0 <= xi < lengh and 0 <= yj < high:
                  count = count + matrix[xi][yj]
      if (count == 3) or (matrix[i][j] == 1 and count == 2):
          return 1
      else:
          return 0

def life(matrix):
    rows, cols = matrix.shape
    result = np.zeros(matrix.shape, dtype=int)
    for i in range(rows):
        for j in range(cols):
            result[i][j] = live_cells(matrix, i, j)
    return result
#a = (np.random.random((5 , 5)) < 0.256).astype(int)
life(a)
```
